```python
options=uc.ChromeOptions()
options.add_argument('--start-maximised')

driver=uc.Chrome(options=options)
url='https://www.meesho.com/sandals-men/pl/3js'
driver.get(url)

time.sleep(5)

#----------------------------Scroll Down Code-------------------------#

def user_pause():
    time.sleep(random.uniform(1.5,3.0))

total_height=driver.execute_script('return document.body.scrollHeight')

cur_position=0
step=400 # itna pixel scroll down

while cur_position<total_height:
    driver.execute_script(f'window.scrollTo(0,{cur_position})')
    user_pause()
    new_height=driver.execute_script('return document.body.scrollHeight')

    if new_height>total_height:
        total_height=new_height
    
    cur_position+=step

time.sleep(5)

html=driver.page_source 

#save file using file handling
with open('meesho_scraped.html','w',encoding='utf-8') as f:
    f.write(html)
print('File Saved Successfully')

code_sample=html[:2500]
print('Code Sample: ', code_sample)
driver.quit()
```

```python
#load file meesho.html
with open('meesho_scraped.html', mode='r', encoding='utf-8') as f:
    html=f.read()
    
#create soup object
soup=BeautifulSoup(html,'lxml')
print('Done')
```

```python
import pandas as pd

# all products
products = soup.find_all(
    'div',
    class_=lambda x: x and 'ProductListItem__GridCol' in x
)

data = []

for product in products:

    # Product Name
    name_tag = product.find(
        'p',
        class_='sc-dOfePm ivsDoe NewProductCardstyled__StyledDesktopProductTitle-sc-6y2tys-5 ffaqxo'
    )

    if name_tag:
        name = name_tag.text.strip()
    else:
        name = "N/A"

    # Price
    price_tag = product.find('h5')

    if price_tag:
        price = price_tag.text.strip()
    else:
        price = "N/A"

    # Rating
    rating_tag = product.find(
        'span',
        class_='sc-dOfePm jklcNf'
    )

    if rating_tag:
        rating = rating_tag.text.strip()
    else:
        rating = "N/A"

    # Reviews
    review_tag = product.find(
        'span',
        class_='sc-dOfePm fTrqZg NewProductCardstyled__RatingCount-sc-6y2tys-22 bzMyLg'
    )

    if review_tag:
        reviews = review_tag.text.strip()
    else:
        reviews = "N/A"

    data.append([name, price, rating, reviews])

# dataframe
df = pd.DataFrame(
    data,
    columns=['Product Name', 'Price', 'Rating', 'Reviews']
)
```
